```python
import random
import time

YELLOW = "\033[93m"
GREEN = "\033[92m"
CYAN = "\033[96m"
RESET = "\033[0m"

# all secret words are Italian foods and drinks
words = ["cappuccino", "lasagne", "spaghetti", "tiramisu",
         "risotto", "ravioli", "gnocchi", "prosciutto",
         "mozzarella", "focaccia", "cannoli", "carbonara"]

def start_game(words):
    """Create a new game state with a random secret word, 6 lives and no guessed letters."""
    # everything the game needs to remember is stored in one dictionary
    game_state = {
        "secret_word": random.choice(words),
        "guessed_letters": set(),   # a set so the same letter can't be stored twice
        "lives": 6,
        "game_over": False
    }
    return game_state
```

```python
def show_status(game_state):
    """Print the remaining lives, the letters guessed so far and the secret word with unguessed letters hidden."""
    # build the word, showing guessed letters and "_" for the rest
    word_display = ""
    for letter in game_state["secret_word"]:
        if letter in game_state["guessed_letters"]:
            word_display += letter + " "
        else:
            word_display += "_ "

    print("\n" + "─" * 40)
    print("Remaining lives: " + str(game_state["lives"]))
    # put the guessed letters in alphabetical order and separate them with commas
    print("Letters guessed: " + ", ".join(sorted(game_state["guessed_letters"])))
    print()
    print(GREEN + "Secret Word: " + word_display + RESET)
    print()
```

```python
def player_guess(game_state):
    """Ask the player for a letter, validate it and update the game state."""
    while True:
        print(f"\n{YELLOW}  >>> Please guess a letter: {RESET}", end="")
        # remove extra spaces and make it lowercase so "A " counts as "a"
        guess = input("").strip().lower()

        # only accept one single letter (no numbers, symbols or empty input)
        if len(guess) != 1 or not guess.isalpha():
            print("Please enter a single valid letter.")
        # a repeated letter doesn't cost a life
        elif guess in game_state["guessed_letters"]:
            print("You already guessed that letter!")
        # wrong letter: save it and lose a life
        elif guess not in game_state["secret_word"]:
            game_state["guessed_letters"].add(guess)
            game_state["lives"] -= 1
            time.sleep(1)
            print("Wrong! The letter is not part of the Secret Word.")
            break
        # right letter: save it so it shows up in the word
        else:
            game_state["guessed_letters"].add(guess)
            time.sleep(1)
            print("Nice! You guessed a letter!")
            break
```

```python
def check_game_over(game_state):
    """Check whether the player has won or lost, and end the game if so."""
    # no lives left means the player lost
    if game_state["lives"] == 0:
        print(CYAN + "GAME OVER! The correct word was " + game_state["secret_word"] + "." + RESET)
        game_state["game_over"] = True
    # if nothing is left, every letter was found and the player won
    elif len(set(game_state["secret_word"]) - game_state["guessed_letters"]) == 0:
        show_status(game_state)
        print(CYAN + "YOU WIN! The correct word was " + game_state["secret_word"] + "." + RESET)
        game_state["game_over"] = True
```

```python
print("Ciao! Welcome to ITALIAN HANGMAN. Your Secret Word is related to Italian cuisine. Try to guess a letter!")
time.sleep(2)

# outer loop: one full game each time the player says yes
play_again = "yes"
while play_again == "yes":
    state = start_game(words)   # new word and 6 lives for every game

    # inner loop: one guess per round until the game is won or lost
    while state["game_over"] == False:
        show_status(state)
        player_guess(state)
        check_game_over(state)

    play_again = input("\nDo you want to play again? (yes/no): ").strip().lower()

print("Grazie for playing ITALIAN HANGMAN!")                          
```
